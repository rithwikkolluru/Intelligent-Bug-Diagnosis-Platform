# Dataset analysis sample notebook

This notebook is designed to run on the shipped fixture data and can be reused with real defect files once they are placed in `data/raw/`.
It intentionally keeps the analysis lightweight and runnable without a full dataset.

```python
from pathlib import Path
import csv
import json

root = Path.cwd().resolve().parent
fixture_dir = root / 'tests' / 'fixtures'
print(f'Fixture directory exists: {fixture_dir.exists()}')
records = []
for path in sorted(fixture_dir.glob('*')):
    if path.suffix.lower() == '.csv':
        with path.open('r', encoding='utf-8') as handle:
            records.extend(list(csv.DictReader(handle)))
    elif path.suffix.lower() in {'.json', '.jsonl'}:
        with path.open('r', encoding='utf-8') as handle:
            if path.suffix.lower() == '.json':
                payload = json.load(handle)
                if isinstance(payload, list):
                    records.extend(payload)
                else:
                    records.append(payload)
            else:
                for line in handle:
                    line = line.strip()
                    if line:
                        records.append(json.loads(line))
print(f'Records loaded: {len(records)}')
print('Columns:', sorted({key for row in records for key in row.keys()}))
```

```python
from collections import Counter
counter = Counter(str(record.get('project') or 'unknown').lower() for record in records)
print(counter)
```
